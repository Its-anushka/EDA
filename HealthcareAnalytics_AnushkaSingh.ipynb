```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# **Data Preparation and Cleaning**

```python
df = pd.read_csv("heart_disease_health_indicators_BRFSS2015 2 (1).csv")
df
```

**Checking for null values**


```python
df.isnull().sum()
```

```python
df.info()
```

Function for mapping values in columns having binary data

```python
df['Sex'] = df['Sex'].map({0: 'Female', 1: 'Male'})
df.head()
```

```python
df['Smoker'] = df['Smoker'].map({0: 'Non-Smoker', 1: 'Smoker'})
df.head()
```

```python
df['Education']= df['Education'].map({1.0 : 'No School', 2.0 : 'Primary School', 3.0 : 'Secondary School', 4.0 : 'High School',5.0 : 'Intermediate', 6.0 : 'Grad/Post-gard'})
df.head()
```

```python
df['HeartDiseaseorAttack']= df['HeartDiseaseorAttack'].map({1 : "With Heart disease", 0 : "Without Heart disease"})
```

```python
df.head()
```

```python
df['HighBP']= df['HighBP'].map({1 : "High Blood Pressure", 0 : "Normal"})
df['HighChol']= df['HighChol'].map({1 : "High Cholesterol", 0 : "Normal"})
df.head()
```

```python

df['Age'] = df['Age'].map({1: '18-24', 2: '25-30', 3: '30-35', 4: '35-40', 5: '40-45', 6: '45-50', 7:'50-55', 8: '55-60', 9: '60-65', 10: '65-70', 11: '70-75',12: '75-80', 13: '80+' })
df.head()
```

```python
#Categorizing BMI values into categories
def categorize_bmi(bmi):
    if bmi < 18.5:
        return 'Underweight'
    elif 18.5 <= bmi < 24.9:
        return 'Normal weight'
    elif 25.0 <= bmi < 29.9:
        return 'Overweight'
    else:
        return 'Obese'

df['Category'] = df['BMI'].map(categorize_bmi)
df['Category']
```

```python
df['NoDocbcCost'] = df['NoDocbcCost'].map({0: 'No', 1: 'Yes'})
df.head()
```

```python
df['Diabetes'] = df['Diabetes'].map({0: 'Non-diabetic', 1: 'Diabetic'})
df.head()
```

### **Relationship between Mental Health Days with BMI (in people with heart disease)**

Indicates the number of days, within the last 30 days where the person had bad mental health

```python

sns.regplot(y="MentHlth", x="BMI", data=df1)
plt.ylabel("Mental Health Days")
plt.xlabel("BMI")
plt.title("Scatter Plot of Mental Health Days vs. BMI ")
plt.show()
```

### **Relationship between Physical Health Days with BMI (in people with heart disease)**

Indicates the number of days, within the last 30 days where the person had bad physical health

```python
sns.regplot(y="PhysHlth", x="BMI", data=df1)
plt.ylabel("Physical Health Days")
plt.xlabel("BMI")
plt.title("Scatter Plot of Physical Health Days vs. BMI")
plt.show()
```

Although the histograms of physical and mental health days depict that they are not having any direct relationship with heart disease, as the least value bin (0-5) has majority of the heart dieases cases, but they do have a linear with BMI which is another common trait leading to heart disease. From this we can infer that Physical and mental health are also an important factor affecting heart health
